```python
import pandas as pd
import numpy as np
import re
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import nltk
```

```python
# Import the CSV in a dataframe
news = pd.read_csv("Dataset/Fake_Dataset.csv", index_col="Unnamed: 0")
print("Imported CSV")
```

```python
# Downloading and displaying the stopwords in English
nltk.download("stopwords")
print(stopwords.words("english"))
```

```python
news.shape
news.head()
```

```python
news.isnull().sum()
```

```python
# Replacing the null values with empty string
news = news.fillna('')
```

```python
# Merging title and text columns
news['content'] = news['title']+''+news['text']
```

```python
# Stemming the content [ Process of removing a word to its root word] eg. actor, actress, acting --> act
port_stem = PorterStemmer()
```

```python
def stemming(content):
    stemmed_content = re.sub('[^a-zA-Z]',' ',content)
    stemmed_content = stemmed_content.lower()
    stemmed_content = stemmed_content.split()
    stemmed_content = [port_stem.stem(word)
                       for word in stemmed_content if not word in stopwords.words('english')]
    stemmed_content = ' '.join(stemmed_content)
    return stemmed_content
```

```python
news['content'] = news['content'].apply(stemming)
```

```python
# Separating the data and the labels
data = news['content'].values
label = news['label'].values
```

```python
# Converting the textual data to numerical data

vectorizer = TfidfVectorizer()

vectorizer.fit(data)
data = vectorizer.transform(data)
```

```python
# Split the data to training and test data
data_train, data_test, label_train, label_test = train_test_split(data, label, test_size = 0.3, stratify = label, random_state =2)
```

```python
# Training the Logistic Regression Model
model = LogisticRegression()
model.fit(data_train, label_train)
```

```python
# Evaluation
# Accuracy Score
data_train_prediction = model.predict(data_train)
training_data_accuracy = accuracy_score(data_train_prediction, label_train)

data_test_prediction = model.predict(data_test)
test_data_accuracy = accuracy_score(data_test_prediction, label_test)

print("Training data accuracy:", training_data_accuracy*100,"%")
print("Test data accuracy:", test_data_accuracy*100,"%")
```

```python
# Making a predictive system
data_new = data_test[0]
prediction = model.predict(data_new)

if (prediction[0] == 0):
    print("The news is real.")
else:
    print("The news is fake!")
```
