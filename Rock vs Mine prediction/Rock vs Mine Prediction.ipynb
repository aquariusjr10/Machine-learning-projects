```python
#Importing Dependencies
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import pickle
```

```python
df = pd.read_csv("Dataset/sonar_data.csv", header=None)
print("Dataset Imported in the variable Data")
```

```python
print("The first five rows of the dataset\n", df.head())
print("\nThe shape of the dataset:", df.shape)
print("\nThe Count of Rocks and Mines:\n", df[60].value_counts())
```

```python
#Separating the data and the labels
data = df.drop(columns=60, axis=1)
label = df[60]
```

```python
#Splitting training and test data
data_train, data_test, label_train, label_test = train_test_split(data, label, test_size = 0.1, stratify = label, random_state = 1)
```

```python
print("Shape of the test train data:", "\n Test Data Shape:", data_test.shape, "\n Train Data Shape:", data_train.shape)
```

```python
#LOGISTIC REGRESSION MODEL
model = LogisticRegression()

#Training the model with the training data
model.fit(data_train, label_train)
```

```python
#Model Evaluation
data_train_prediction = model.predict(data_train)
data_test_prediction = model.predict(data_test)
training_data_accuracy = accuracy_score(data_train_prediction, label_train)
test_data_accuracy = accuracy_score(data_test_prediction, label_test)
print("Accuracy Score for training data:", training_data_accuracy*100,"%")
print("Accuracy Score for test data:", test_data_accuracy*100,"%")
```
